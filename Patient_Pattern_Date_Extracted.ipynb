```python
fake_note = pymupdf.open("/content/Progress notes.pdf")

def reg_pattern_find(pattern, text, page_number):
  match = re.search(pattern, text, re.IGNORECASE)
  if match:
    return match.group(1)
  else:
    print(f"Page {page_number + 1} does not contain the pattern, please recheck this page content")
    return None

for page_number, page in enumerate(fake_note):
  text = page.get_text()
  date_data = {
      'Created Date': [reg_pattern_find(r"(?:Created Date):\s*(\d{2}/\d{2}/\d{4})", text, page_number)],
      'Effective Date': [reg_pattern_find(r"Effective Date:\s*(\d{2}/\d{2}/\d{4})", text, page_number)],
      'Created Date Time': [reg_pattern_find(r"Created Date:\s*\d{2}/\d{2}/\d{4}\s*(\d{2}:\d{2})", text, page_number)],
      'Effecitve Date Time': [reg_pattern_find(r"Effective Date:\s*\d{2}/\d{2}/\d{4}\s*(\d{2}:\d{2})", text, page)]

  }
  date_df = pd.DataFrame(date_data)
date_df
```

```python
#Check for row index with null values:
row_with_null = date_df[date_df.isnull().any(axis=1)]
print(row_with_null)
```
